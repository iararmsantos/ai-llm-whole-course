```python
import gradio as gr
from transformers import pipeline

# Load Whisper model (you can choose small, medium, large, etc.)
asr = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small.en",
    return_timestamps=False
)

def transcribe(audio_file):
    if audio_file is None:
        return "Please upload an audio file."

    # audio_file is a path string created by Gradio
    result = asr(audio_file)
    return result["text"]

with gr.Blocks() as demo:
    gr.Markdown("## 🎤 Audio Transcription Chat")

    with gr.Row():
        audio_input = gr.Audio(
            label="Upload audio file",
            type="filepath"
        )

    with gr.Row():
        transcribe_btn = gr.Button("Transcribe")

    chatbox = gr.Chatbot(label="Transcription Output")
    
    def handle_transcription(history, audio_file):
        text = transcribe(audio_file)

        history = history or []
        history.append({"role": "user", "content": "Transcribe this audio"})
        history.append({"role": "assistant", "content": text})

        return history

    transcribe_btn.click(
        fn=handle_transcription,
        inputs=[chatbox, audio_input],
        outputs=chatbox
    )

demo.launch()
```
