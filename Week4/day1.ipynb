The Chinchilla Scaling Law
- Number of parameters ~ proportional to the number of training tokens
    - if you upgrade to a model with double the number of weights, this law suggests you need twice data and vice-versa

- as much parameter you have more training tokens you have to have

Limitations of BEchmarks
- Training Data Contamination (Training the LLM to pass the test)
- Not consistently applied
- Too narrow in scope
- Hard to measure nuanced reasoning
- saturation: first tests tested excellence then other tests doesn't reduce this numbers
- Overfitting: you consistently pick a model you train this model
- Frontier LLMs may be aware that they are being evaluated
